numeric variables that have the strongest relationship with SalePrice:

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

correlations = numeric_df.corr()['SalePrice'].sort_values(ascending=False)

correlations.head(15)
```

Correlation Analysis:

The correlation analysis identified several numerical variables that have strong positive relationships with SalePrice. OverallQual showed the highest correlation with the target (r = 0.791), followed by GrLivArea (r = 0.709). Garage capacity (GarageCars, r = 0.640), garage area (GarageArea, r = 0.623), total basement area (TotalBsmtSF, r = 0.614), and first-floor area (1stFlrSF, r = 0.606) also showed notable positive relationships. These findings suggest that overall construction quality, usable living space, garage capacity, basement area, and property age are potentially important factors in explaining house prices.

```python
plt.figure(figsize=(9, 6))

df.boxplot(column='SalePrice', by='OverallQual')

plt.xlabel('Overall Quality Rating')
plt.ylabel('Sale Price')
plt.title('SalePrice by Overall Quality')
plt.suptitle('')

plt.show()
```

Overall Quality Analysis:

The boxplot shows a strong positive relationship between OverallQual and SalePrice. Properties with higher overall quality ratings generally have substantially higher median sale prices. The spread in sale prices also increases for higher quality ratings, indicating greater price variation among high-quality properties. This supports the correlation result where OverallQual had the strongest positive correlation with SalePrice (r = 0.791).

```python
neighborhood_prices = (
    df.groupby('Neighborhood')['SalePrice']
      .median()
      .sort_values()
)

plt.figure(figsize=(12, 6))
neighborhood_prices.plot(kind='bar')

plt.xlabel('Neighborhood')
plt.ylabel('Median Sale Price')
plt.title('Median SalePrice by Neighborhood')
plt.xticks(rotation=90)

plt.show()
```

Neighborhood Analysis:

Median sale prices vary considerably across neighborhoods, indicating that location is an important factor in property valuation. Neighborhoods such as NridgHt, NoRidge, and StoneBr show notably higher median sale prices, while MeadowV, IDOTRR, and BrDale have substantially lower median prices. This suggests that Neighborhood may be an important categorical predictor of house price.

Outlier Analysis:

Boxplots of SalePrice and GrLivArea indicate the presence of several high-value and large-area observations.
```python
# Check columns with very low variation
low_variation = []

for col in df.columns:
    if df[col].nunique(dropna=False) <= 2:
        low_variation.append((col, df[col].nunique(dropna=False)))

low_variation
```

```python
for col in ['Street', 'Utilities', 'CentralAir']:
    print(f"\n{col}")
    print(df[col].value_counts())
    print(df[col].value_counts(normalize=True) * 100)
```

```python
dominant_features = []

for col in df.columns:
    counts = df[col].value_counts(normalize=True, dropna=False)

    if len(counts) > 0:
        dominant_pct = counts.iloc[0] * 100

        if dominant_pct >= 95:
            dominant_features.append(
                (col, round(dominant_pct, 2))
            )

dominant_features
```

Low-Variation Feature Analysis:

Several variables were found to be strongly dominated by a single value or category. For example, Street is recorded as Pave for approximately 99.59% of properties, while Utilities is AllPub for approximately 99.93% of observations. Other variables such as Condition2, RoofMatl, Heating, LowQualFinSF, 3SsnPorch, PoolArea, PoolQC, MiscFeature, and MiscVal also show very high dominance by one value. These variables may provide limited discriminatory information for predictive modelling and should be reviewed during preprocessing and feature-selection stages rather than removed automatically.

Data Quality Analysis:

The dataset was examined for missing values, duplicate rows, outliers, low-variation variables, and internal consistency. No duplicate rows were found. Several features contain missing values, but the consistency checks indicate that many of these values are meaningful rather than erroneous. For example, properties with missing garage-related categorical fields such as GarageType also have GarageCars = 0 and GarageArea = 0, indicating the absence of a garage. Similarly, properties with missing basement quality fields have BsmtFinSF1, BsmtFinSF2, BsmtUnfSF, and TotalBsmtSF equal to zero, indicating the absence of a basement. Therefore, these missing values should be treated as structural absence during preprocessing rather than automatically replaced as unknown data. The dataset also contains several potential outliers in SalePrice and GrLivArea, as well as some low-variation features such as Street and Utilities. These issues should be reviewed carefully during the preprocessing stage.
